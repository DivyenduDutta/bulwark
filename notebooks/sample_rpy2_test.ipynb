Sample code to send a numpy array to rpy2 and get back its mean

```python
import numpy as np
from rpy2.robjects.packages import importr
from rpy2.robjects import default_converter, numpy2ri
from rpy2.robjects.conversion import localconverter
```

Use [this](https://rpy2.github.io/doc/v3.5.x/html/robjects_convert.html#conversion) in case there's an error `Conversion rules for rpy2.robjects appear to be missing`

```python
a = np.array([1, 2, 3])
print(a.shape)
print("Mean using numpy: ", np.mean(a))

base = importr("base")
with localconverter(default_converter + numpy2ri.converter):
    r_mean = base.mean(a)
mean_val = float(r_mean[0])
print("Mean using rpy2: ", mean_val)
```
